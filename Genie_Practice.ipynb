```python
%sql
-- Create a sample customers table
CREATE TABLE IF NOT EXISTS workspace.default.genie_customers (
    customer_id INT COMMENT 'Unique identifier for each customer',
    customer_name STRING COMMENT 'Full name of the customer',
    country STRING COMMENT 'Two-letter country code, e.g., US, CA, IN'
);

-- Create a sample sales table
CREATE TABLE IF NOT EXISTS workspace.default.genie_sales (
    order_id INT COMMENT 'Unique identifier for the order',
    customer_id INT COMMENT 'Maps to genie_customers.customer_id',
    order_date DATE COMMENT 'The date the transaction happened',
    amount DOUBLE COMMENT 'The total USD value of the purchase',
    status STRING COMMENT 'Can be Completed, Pending, or Returned'
);

-- Insert a quick row of mock data for testing
INSERT INTO workspace.default.genie_customers VALUES (101, 'Alice Smith', 'US');
INSERT INTO workspace.default.genie_sales VALUES (9001, 101, '2026-10-01', 150.00, 'Completed');
```
