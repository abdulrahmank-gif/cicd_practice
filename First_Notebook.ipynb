```python
# COMMAND ----------
# 1. Import necessary PySpark functions
from pyspark.sql import functions as F
from pyspark.sql.types import DoubleType

# COMMAND ----------
# 2. Create sample mock data (Simulating your raw data pipeline)
data = [
    ("Product_A", 10.0),
    ("Product_B", 100.0),
    ("Product_C", 1000.0),
    ("Product_D", 50.5)
]

columns = ["product_name", "revenue"]
df = spark.createDataFrame(data, schema=columns)

print("Original Dataframe:")
df.show()

# COMMAND ----------
# 3. Define and apply the Log Transformation (Handling log(0) safety)
# We add 1 to the revenue to prevent log(0) mathematical errors if revenue is 0.
df_transformed = df.withColumn("log_transformed_revenue", F.log(F.col("revenue") + 1))

print("Transformed Dataframe:")
df_transformed.show()

# COMMAND ----------
# 4. Save the transformed data to a Delta Table (Production Target)
# In production, change the path or table name as needed
df_transformed.write \
    .format("delta") \
    .mode("overwrite") \
    .saveAsTable("workspace.default.transformed_revenue_metrics")

print("Pipeline executed and Delta table updated successfully!")
```
